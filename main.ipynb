# Computer Hardware Benchmarking using Statistical Learning Approaches
## Mark Richards' Pipes & Filters Software Architecture + Production MLOps

### Architectural Highlights:
1. **Pipes & Filters Pattern**: Every processing stage is a modular `Filter` connected by `Pipe` payload channels (`PipelineContext`).
2. **Strategy & Factory Pattern**: Classical ML algorithm strategies (`LinearRegressionStrategy`, `RandomForestStrategy`, `ExtraTreesStrategy`, etc.) instantiated dynamically via `ModelFactory`.
3. **Production REST API**: FastAPI server hosting real-time inference (`/predict/cpu`, `/predict/gpu`) and data ingestion triggers (`/ingest`).
4. **Docker Containerization**: Dockerfile and Docker Compose setup for containerized deployment.

--- 
### Section 1: Environment & Pipes & Filters Imports

```python
import pandas as pd
import numpy as np
from src.architecture import Pipeline, PipelineContext
from src.architecture.filters import (
    ValidationFilter,
    EDAFilter,
    DataCleaningFilter,
    FeatureEngineeringFilter,
    ModelTrainingFilter,
    ModelEvaluationFilter,
    RegistryPromotionFilter,
)
from src.models.strategies import ModelFactory

print("Mark Richards Pipes & Filters framework ready.")
print("Registered Strategies:", ModelFactory.get_registered_names())
```

--- 
### Section 2: Composing & Executing the Intel CPU Pipeline Chain

```python
cpu_pipeline = (
    ValidationFilter()
    | EDAFilter()
    | DataCleaningFilter()
    | FeatureEngineeringFilter()
    | ModelTrainingFilter(strategies=ModelFactory.get_registered_names())
    | ModelEvaluationFilter(min_r2=0.70, max_rmse=0.50)
    | RegistryPromotionFilter()
)

cpu_ctx = PipelineContext(domain_key="cpu", raw_filepath="Intel_CPUs.csv", target_col="Processor_Base_Frequency")
cpu_res = cpu_pipeline.execute(cpu_ctx)

print(f"Champion Strategy: {cpu_res.champion_model_name}")
print(f"Champion Out-of-sample Test R²: {cpu_res.champion_metrics['R2_Score']:.4f}")
```

--- 
### Section 3: Composing & Executing the All GPU Pipeline Chain

```python
gpu_pipeline = (
    ValidationFilter()
    | EDAFilter()
    | DataCleaningFilter()
    | FeatureEngineeringFilter()
    | ModelTrainingFilter(strategies=ModelFactory.get_registered_names())
    | ModelEvaluationFilter(min_r2=0.60, max_rmse=300.0)
    | RegistryPromotionFilter()
)

gpu_ctx = PipelineContext(domain_key="gpu", raw_filepath="All_GPUs.csv", target_col="Core_Speed")
gpu_res = gpu_pipeline.execute(gpu_ctx)

print(f"Champion Strategy: {gpu_res.champion_model_name}")
print(f"Champion Out-of-sample Test R²: {gpu_res.champion_metrics['R2_Score']:.4f}")
```
