# 03 — Results report

A lightweight, reproducible report based on registered production candidates—not an independently trained notebook model.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'models').exists():
    ROOT = ROOT.parent

registry = json.loads((ROOT / 'models/registry.json').read_text())
summary = pd.DataFrame([
    {'domain': domain, 'model': value['model_name'], **value['metrics'], 'updated_at': value['updated_at']}
    for domain, value in registry.items()
])
summary
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid', palette='deep')
ax = sns.barplot(data=summary, x='domain', y='R2_Score', hue='model', legend=False)
ax.set(title='Registered champion held-out R²', ylabel='R² score', xlabel='Domain')
plt.ylim(0, 1)
```
