# 01 — Exploratory Data Analysis 

Explore the immutable raw inputs. This notebook is intentionally read-only: data cleaning and feature engineering live in `src/benchmarking/`.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent

cpu = pd.read_csv(ROOT / 'data/raw/Intel_CPUs.csv')
gpu = pd.read_csv(ROOT / 'data/raw/All_GPUs.csv')
sns.set_theme(style='whitegrid', palette='muted')
cpu.shape, gpu.shape
```

```python
display(cpu.head())
display(gpu.head())
missingness = pd.DataFrame({
    'cpu_missing_pct': cpu.isna().mean().sort_values(ascending=False).head(12),
    'gpu_missing_pct': gpu.isna().mean().sort_values(ascending=False).head(12),
})
missingness
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(cpu['Processor_Base_Frequency'].dropna(), ax=axes[0], color='#2b5c8f')
axes[0].set(title='CPU base-frequency distribution', xlabel='GHz')
sns.histplot(gpu['Core_Speed'].dropna(), ax=axes[1], color='#27ae60')
axes[1].set(title='GPU core-speed distribution', xlabel='MHz')
plt.tight_layout()
```

## Guardrail

Do not create `Core_Speed / Memory_Speed` as an input feature: `Core_Speed` is the GPU target and this would leak the answer into training.
